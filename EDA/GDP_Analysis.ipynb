```python
import os
import plotly.express as px
import pandas as pd
import pathlib
import numpy as np
import matplotlib.pyplot as plt
```

```python
#path = pathlib.Path() / "Datasets" / "GDP" / "API_SSD_DS2_en_csv_v2_2529.csv"

df = pd.read_csv("./Datasets/GDP/API_SSD_DS2_en_csv_v2_2529.csv", skiprows=4)
df.head(20)
```

```python
indic = df["Indicator Name"][3]
indic
```

```python
indic2 = df["Indicator Name"][4]
indic2
```

```python
df_copy = df.copy()

indicators = df_copy["Indicator Name"].value_counts().reset_index()
indicators.columns = ["Indicator Types", "Total Amount"]

fig = px.bar(indicators, x="Indicator Types", y="Total Amount", text_auto=True, title="Total Amount of Indicators per Type")
fig.update_layout(yaxis={'categoryorder':'total ascending'}, yaxis_range=[0, indicators["Total Amount"].max()*1.25])
fig.update_traces(textposition='outside')
fig.show()
```

```python
df_copy2 = df.copy()

indicators2 = df_copy2["Indicator Code"].value_counts().reset_index()
indicators2.columns = ["Indicator Types", "Total Amount"]

fig = px.bar(indicators2, x="Indicator Types", y="Total Amount", text_auto=True, title="Total Amount of Indicators per Type")
fig.update_layout(yaxis={'categoryorder':'total ascending'}, yaxis_range=[0, indicators2["Total Amount"].max()*1.25])
fig.update_traces(textposition='outside')
fig.show()
```
