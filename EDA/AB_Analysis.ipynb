```python
import plotly.express as px

gdf2_loc = gdf2.copy()

gdf2_clean = gdf2_loc.dropna(subset=['center_lat', 'center_lon']) #handling NaNs

fig = px.density_map(
    gdf2_clean, 
    lat='center_lat', 
    lon='center_lon', 
    radius=7,
    opacity=0.95,
    center=dict(lat=gdf2_clean.center_lat.mean(), lon=gdf2_clean.center_lon.mean()), #Without a defined center, most mapping libraries have no idea where your data is located.
    zoom=11,
    map_style='open-street-map', 
    height=800,
    color_continuous_scale="Viridis",
    title=""
)

fig.show()
```
