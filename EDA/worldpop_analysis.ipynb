```python
from pathlib import Path
import tifffile
import numpy as np
import rasterio
from rasterio.plot import show, show_hist
```

```python
folder = Path("./Datasets/worldpop")

for file_path in folder.glob("*tif*"):
    print(f"Reading: {file_path.name}")
    data=tifffile.imread(file_path)
    print(f"Shape:{data.shape}, Dtype:{data.dtype}")
```

```python
from matplotlib import pyplot

for file_path in folder.glob("*tif*"):
    print(f"Reading: {file_path.name}")
    with rasterio.open(file_path) as src:
        data = src.read(1)
        pyplot.imshow(data, cmap="pink")
        pyplot.show()
```

```python
for file_path in folder.glob("*tif*"):
    print(f"Reading: {file_path.name}")
    with rasterio.open(file_path) as src:
        data = src.read(1)
        print(f"Shape: {data.shape}, Dtype: {data.dtype}, CRS: {src.crs}")
        show(src)
        show(src.read(), transform = src.transform)
```

```python
 for file_path in folder.glob("*tif*"):
    print(f"Reading: {file_path.name}")
    with rasterio.open(file_path) as src:
        data = src.read(1)
        print(f"Shape: {data.shape}, Dtype: {data.dtype}, CRS: {src.crs}")
        show_hist(src, bins = 50, lw=0.0, stacked=False, alpha=0.3, histtype="stepfilled", title="Histogram for South Sudan")
```

```python
 for file_path in folder.glob("*tif*"):
    print(f"Reading: {file_path.name}")
    with rasterio.open(file_path) as src:
        fig, (axrgb, axhist) = pyplot.subplots(1, 2, figsize=(14,7))
        show(src, ax=axrgb)
        show_hist(src, bins = 50, lw=0.0, stacked=False, alpha=0.3, histtype="stepfilled", title="Histogram", ax=axhist)
        pyplot.show()
```
